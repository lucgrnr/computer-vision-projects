Your filled test dataframe (during Sect. 2 and Sect. 3) must be converted to a submission.csv with two rows per example (one for classification and one for segmentation) and with only a single prediction column (the multi-class/label predictions running length encoded).
```python
def _rle_encode(img):
    """
    Kaggle requires RLE encoded predictions for computation of the Dice score (https://www.kaggle.com/lifa08/run-length-encode-and-decode)

    Parameters
    ----------
    img: np.ndarray - binary img array
    
    Returns
    -------
    rle: String - running length encoded version of img
    """
    pixels = img.flatten()
    pixels = np.concatenate([[0], pixels, [0]])
    runs = np.where(pixels[1:] != pixels[:-1])[0] + 1
    runs[1::2] -= runs[::2]
    rle = ' '.join(str(x) for x in runs)
    return rle

def generate_submission(df):
    """
    Make sure to call this function once after you completed Sect. 2 and Sect. 3! It transforms and writes your test dataframe into a submission.csv file.
    
    Parameters
    ----------
    df: pd.DataFrame - filled dataframe that needs to be converted
    
    Returns
    -------
    submission_df: pd.DataFrame - df in submission format.
    """
    df_dict = {"Id": [], "Predicted": []}
    for idx, _ in df.iterrows():
        df_dict["Id"].append(f"{idx}_classification")
        df_dict["Predicted"].append(_rle_encode(np.array(df.loc[idx, labels])))
        df_dict["Id"].append(f"{idx}_segmentation")
        df_dict["Predicted"].append(_rle_encode(np.array([df.loc[idx, "seg"] == j + 1 for j in range(len(labels))])))
    
    submission_df = pd.DataFrame(data=df_dict, dtype=str).set_index("Id")
    submission_df.to_csv("submission.csv")
    return submission_df
```

# 1. Image classification
The goal here is simple: implement a classification model and train it to recognise all 20 classes (and/or background) using the training set and compete on the test set (by filling in the classification columns in the test dataframe).

```python
class RandomClassificationModel:
    """
    Random classification model: 
        - generates random labels for the inputs based on the class distribution observed during training
        - assumes an input can have multiple labels
    """
    def fit(self, X, y):
        """
        Adjusts the class ratio variable to the one observed in y. 

        Parameters
        ----------
        X: list of arrays - n x (height x width x 3)
        y: list of arrays - n x (nb_classes)

        Returns
        -------
        self
        """
        self.distribution = np.mean(y, axis=0)
        print("Setting class distribution to:\n{}".format("\n".join(f"{label}: {p}" for label, p in zip(labels, self.distribution))))
        return self
        
    def predict(self, X):
        """
        Predicts for each input a label.
        
        Parameters
        ----------
        X: list of arrays - n x (height x width x 3)
            
        Returns
        -------
        y_pred: list of arrays - n x (nb_classes)
        """
        np.random.seed(0)
        return [np.array([int(np.random.rand() < p) for p in self.distribution]) for _ in X]
    
    def __call__(self, X):
        return self.predict(X)
    
model = RandomClassificationModel()
model.fit(train_df["img"], train_df[labels])
test_df.loc[:, labels] = model.predict(test_df["img"])
test_df.head(1)
```

```python
# ─────────────────────────────────────────────────────────────
# 1.1.3 — Label cardinality + positive/negative pair imbalance
# ─────────────────────────────────────────────────────────────
labels_per_image = train_df[labels].sum(axis=1)

print("Label cardinality (number of labels per image):")
print(labels_per_image.describe())
print()
card_counts = labels_per_image.value_counts().sort_index()
for k, v in card_counts.items():
    print(f"  {int(k)} label(s) : {v} images  ({100*v/len(train_df):.1f}%)")

fig, ax = plt.subplots(figsize=(8, 3.5))
ax.bar(card_counts.index, card_counts.values, color="steelblue")
ax.set_xlabel("Labels per image"); ax.set_ylabel("Count")
ax.set_title("Label cardinality distribution")
ax.set_xticks(card_counts.index)
plt.tight_layout(); plt.show()

# Positive / negative (image,class) pair imbalance — drives the loss choice.
total_pairs = len(train_df) * len(labels)
pos_pairs   = int(train_df[labels].values.sum())
neg_pairs   = total_pairs - pos_pairs
print(f"\nPositive (image,class) pairs: {pos_pairs} / {total_pairs} = {100*pos_pairs/total_pairs:.2f}%")
print(f"Negative (image,class) pairs: {neg_pairs} / {total_pairs} = {100*neg_pairs/total_pairs:.2f}%")
print(f"Positive/Negative ratio: 1 : {neg_pairs/pos_pairs:.1f}")

# Implication: most images carry only 1-3 of the 20 classes → ~93-95% of
# (image, class) pairs are NEGATIVE. Plain BCE wastes capacity learning to
# say "no" ~14× more often than "yes". Asymmetric Loss (ASL, Ridnik 2021)
# explicitly down-weights easy negatives via a separate γ⁻ exponent and
# probability-shift; this is the documented "improvement" we apply in 1.X.
```

```python
# ─────────────────────────────────────────────────────────────
# 1.1.4 — Image pixel intensity statistics vs ImageNet
# ─────────────────────────────────────────────────────────────
SAMPLE = min(200, len(train_df))            # full scan would be slow + redundant
rng    = np.random.default_rng(SEED)
idxs   = rng.choice(len(train_df), SAMPLE, replace=False)

means, stds = [], []
for i in idxs:
    img = train_df.iloc[i]["img"].astype(np.float32) / 255.0
    means.append(img.reshape(-1, 3).mean(axis=0))
    stds.append( img.reshape(-1, 3).std(axis=0))

means = np.stack(means)
stds  = np.stack(stds)
dataset_mean = means.mean(axis=0)
dataset_std  = stds.mean(axis=0)

print(f"VOC sample ({SAMPLE} images) per-channel mean: "
      f"R={dataset_mean[0]:.4f}  G={dataset_mean[1]:.4f}  B={dataset_mean[2]:.4f}")
print(f"VOC sample ({SAMPLE} images) per-channel std : "
      f"R={dataset_std[0]:.4f}  G={dataset_std[1]:.4f}  B={dataset_std[2]:.4f}\n")
print(f"ImageNet reference  mean: R={IMG_MEAN[0]:.4f}  G={IMG_MEAN[1]:.4f}  B={IMG_MEAN[2]:.4f}")
print(f"ImageNet reference  std : R={IMG_STD[0]:.4f}  G={IMG_STD[1]:.4f}  B={IMG_STD[2]:.4f}")
print(f"\n|delta(mean)| max: {np.abs(dataset_mean - np.array(IMG_MEAN)).max():.4f}")
print(f"|delta(std)|  max: {np.abs(dataset_std  - np.array(IMG_STD)).max():.4f}")

channels = ["R", "G", "B"]
colors   = ["#e74c3c", "#27ae60", "#2980b9"]
fig, axes = plt.subplots(1, 3, figsize=(14, 3.5))
for i, (ch, col) in enumerate(zip(channels, colors)):
    axes[i].hist(means[:, i], bins=25, color=col, alpha=0.7, label="per-image mean")
    axes[i].axvline(IMG_MEAN[i], color="black", linestyle="--", linewidth=1.2,
                    label=f"ImageNet ({IMG_MEAN[i]:.3f})")
    axes[i].set_title(f"Channel {ch}")
    axes[i].legend(fontsize=8)
plt.suptitle(f"Per-image channel-mean distribution ({SAMPLE} train images) vs ImageNet")
plt.tight_layout(); plt.show()

# Implication: VOC2009 channel statistics are within ~0.02 of ImageNet's
# reference values. So (a) ImageNet IMG_MEAN/IMG_STD is the right
# normalisation — no need to compute dataset-specific stats; and (b) the
# low-level features ImageNet-pretrained ResNet-50 learned should transfer
# to VOC essentially out-of-the-box, justifying the transfer-learning model.
```

| Per-class frequencies are highly skewed: `person` 27.6% vs `sheep` 3.6% — almost 8× ratio | A single global threshold of 0.5 over-biases majority classes. We tune **per-class** thresholds on the val set (cell 1.6) maximising val Dice. |
| Most images carry 1-3 labels of 20 → only **7.13%** of (image, class) pairs are positive (1:13 imbalance) | Plain BCE is fine as a baseline but spends most capacity on easy negatives. **Asymmetric Loss (ASL, Ridnik 2021)** is the documented "improvement" we apply in 1.X — it has a separate γ⁻ ≫ γ⁺ that down-weights easy negatives. |
| Per-channel pixel mean / std are within ~0.02 of ImageNet reference values | Use `IMG_MEAN` / `IMG_STD` for normalisation — no need for dataset-specific stats. ImageNet features should transfer well → motivates the ResNet-50 transfer model. |
| `MultilabelStratifiedKFold` (n=5, fold 0, seed 42) keeps train↔val class-frequency drift very small | A reliable val split lets us tune per-class thresholds without leakage and compare model variants on a fixed held-out set. We use **fold 0** consistently for all of Section 1 (and reuse the same split in Sections 2 & 3 for consistency). |

With the design space pinned down, the next subsections build the data
pipeline (1.2), define the two architectures (1.3), the loss (1.4), the
training loop (1.5), and per-class threshold tuning (1.6).


## 1.2 Data pipeline — Dataset, augmentations, DataLoaders

Three pieces:

1. **`VOCClassificationDataset`** — a thin in-memory wrapper over `train_df`. Each sample yields one image (H×W×3 uint8) and its 20-bit float label vector. Reading directly from the already-in-RAM dataframe avoids per-step disk I/O.
2. **Albumentations transforms** — train and val pipelines using a single, well-maintained augmentation library that we'll re-use for segmentation (Section 2). Both pipelines end with `ImageNet Normalize` + `ToTensorV2` so the output is a `C×H×W float` tensor ready for the model.
   - **Train**: `RandomResizedCrop(224)` (random scale + aspect — handles 1.1.1's varied sizes/aspects), `HorizontalFlip(p=0.5)` (every VOC class is left-right symmetric except text on tvmonitors, which is rare), `ColorJitter` (mild brightness / contrast / saturation / hue — robustness to camera + lighting differences).
   - **Val/Test**: deterministic `Resize(224, 224)` only — no randomness, so per-class thresholds tuned on val (1.6) are reproducible at inference time.
3. **DataLoaders** — built from the same fold-0 split we sanity-checked in 1.1.5. `batch_size=32` is comfortable for 224×224 RGB on a typical GPU. The `SMOKE` flag from the setup cell shrinks the train set to 50 images so the entire notebook runs end-to-end in seconds during plumbing checks.

```python
# ─────────────────────────────────────────────────────────────
# 1.2.1 — VOCClassificationDataset
# Caches images/targets as lists/arrays so __getitem__ avoids pandas iloc
# overhead during every epoch. If label_cols=None, the same class is used
# for test-time inference and yields only image tensors.
# ─────────────────────────────────────────────────────────────

class VOCClassificationDataset(Dataset):
    """Yields image tensors, and optionally 20-bit float labels."""
    def __init__(self, df: pd.DataFrame, label_cols=None, transform=None):
        self.images = df["img"].to_list()
        self.label_cols = None if label_cols is None else list(label_cols)
        self.targets = (None if label_cols is None
                        else df[self.label_cols].to_numpy(dtype=np.float32))
        self.transform = transform

    def __len__(self):
        return len(self.images)

    def __getitem__(self, idx):
        img = self.images[idx]                                    # H×W×3 uint8
        if self.transform is not None:
            img = self.transform(image=img)["image"]              # → C×H×W float tensor
        if self.targets is None:
            return img
        return img, torch.from_numpy(self.targets[idx])
```

```python
# ─────────────────────────────────────────────────────────────
# 1.2.2 — Albumentations transforms
#   Train: RandomResizedCrop (random scale + aspect = free augmentation —
#          motivated by 1.1.1's wide aspect-ratio range), HorizontalFlip
#          (every VOC class is left-right symmetric), mild ColorJitter
#          (robustness to lighting / camera differences).
#   Val/Test: deterministic Resize(224, 224) — must be reproducible because
#          we tune per-class thresholds on val output (1.6).
#   Both end with ImageNet Normalize (justified by 1.1.4) + ToTensorV2.
# ─────────────────────────────────────────────────────────────

IMG_SIZE = 224

cls_train_tf = A.Compose([
    A.RandomResizedCrop(size=(IMG_SIZE, IMG_SIZE), scale=(0.6, 1.0),
                        ratio=(0.75, 1.3333), p=1.0),
    A.HorizontalFlip(p=0.5),
    A.ColorJitter(brightness=0.2, contrast=0.2, saturation=0.2, hue=0.05, p=0.5),
    A.Normalize(mean=IMG_MEAN, std=IMG_STD),
    ToTensorV2(),
])

cls_val_tf = A.Compose([
    A.Resize(IMG_SIZE, IMG_SIZE),
    A.Normalize(mean=IMG_MEAN, std=IMG_STD),
    ToTensorV2(),
])

# Sanity: run each transform on one image and inspect shape / dtype / range
sample_img = train_df.iloc[0]["img"]
out_train  = cls_train_tf(image=sample_img)["image"]
out_val    = cls_val_tf  (image=sample_img)["image"]

print(f"Sample raw    : {sample_img.shape}  uint8  [{sample_img.min()}, {sample_img.max()}]")
print(f"After train TF: {tuple(out_train.shape)}  {out_train.dtype}  "
      f"[{out_train.min():.3f}, {out_train.max():.3f}]")
print(f"After val   TF: {tuple(out_val.shape)}  {out_val.dtype}  "
      f"[{out_val.min():.3f}, {out_val.max():.3f}]")
```

## 1.3 Architectures — two models, same task

The PDF (Sect. 2.1) explicitly encourages comparing **(i) a model trained from scratch** with **(ii) a model using transferred pre-trained weights + fine-tuning**. We do exactly that:

| Model | Style | Backbone | Head | ~params | Why |
|---|---|---|---|---|---|
| **SmallCNN** | from scratch | 5 conv blocks (32→64→128→256→512), each Conv→BN→ReLU→MaxPool | GAP → Linear(512, 20) | ~1.6 M | Lower bound — what can be learned from ~600 images with no external supervision. Useful baseline for comparison. |
| **ResNet50Cls** | transfer | ResNet-50, ImageNet `IMAGENET1K_V2` weights | Linear(2048, 512) → ReLU → Dropout(0.2) → Linear(512, 20) | ~25 M | The realistic model — most of the heavy lifting (low-level features) is already done by the ImageNet backbone. Fine-tuning + a tiny head should converge fast on ~600 images. |

**Weight provenance** — `IMAGENET1K_V2` are torchvision's improved ResNet-50 weights, trained by Meta on ILSVRC 2012 (ImageNet) using a recipe described in the torchvision blog post. ImageNet has **no PASCAL VOC overlap** beyond what's allowed by the assignment, so this satisfies the hard constraint "No VOC-pretrained weights".

Both models output **20 raw logits** — we keep the sigmoid inside `BCEWithLogitsLoss` / `AsymmetricLossOptimized` (1.4 / 1.X) for numerical stability.

`ResNet50Cls.param_groups()` returns two AdamW parameter groups so the optimiser can use a lower learning rate (1e-4) for the pretrained backbone than the new head (1e-3) — a standard fine-tuning trick that prevents the random head from destabilising the well-conditioned features.

```python
# ─────────────────────────────────────────────────────────────
# 1.3a — SmallCNN (trained from scratch)
# Five conv blocks doubling channels (32 → 512), each block:
#   Conv 3×3 (padding=1, no bias) → BatchNorm → ReLU → MaxPool 2×2
# After 5 blocks the spatial dim collapses 224 → 7. Global Average
# Pool turns the 512-channel 7×7 feature map into a 512-vector;
# final Linear emits 20 logits. BatchNorm is critical — without it,
# training a 1.6 M-param net from scratch on ~600 images is unstable.
# ─────────────────────────────────────────────────────────────

class SmallCNN(nn.Module):
    """5-block VGG-style CNN, GAP head, 20-logit output. From-scratch baseline."""
    def __init__(self, n_classes=20, in_ch=3):
        super().__init__()
        chs = [32, 64, 128, 256, 512]
        layers = []
        prev = in_ch
        for c in chs:
            layers += [
                nn.Conv2d(prev, c, kernel_size=3, padding=1, bias=False),
                nn.BatchNorm2d(c),
                nn.ReLU(inplace=True),
                nn.MaxPool2d(2),
            ]
            prev = c
        self.features = nn.Sequential(*layers)
        self.gap     = nn.AdaptiveAvgPool2d(1)
        self.head    = nn.Linear(chs[-1], n_classes)

    def forward(self, x):
        x = self.features(x)            # (B, 512, 7, 7) for 224×224 input
        x = self.gap(x).flatten(1)      # (B, 512)
        return self.head(x)             # (B, n_classes)
```

```python
# ─────────────────────────────────────────────────────────────
# 1.3b — ResNet50Cls (transfer learning from ImageNet)
# Backbone: torchvision ResNet-50 with IMAGENET1K_V2 weights
# (ILSVRC 2012 / ImageNet — no VOC overlap, satisfies hard constraint).
# Replace the original 1000-class fc with a 512-unit bottleneck +
# Dropout(0.2) + 20-class head. param_groups() returns separate
# parameter groups so the optimiser can apply a lower LR (1e-4) to
# the pretrained backbone than to the new randomly-init head (1e-3).
# ─────────────────────────────────────────────────────────────

class ResNet50Cls(nn.Module):
    """ResNet-50 (ImageNet-pretrained) backbone + 20-logit head."""
    def __init__(self, n_classes=20, dropout=0.2, pretrained=True):
        super().__init__()
        weights = ResNet50_Weights.IMAGENET1K_V2 if pretrained else None
        self.backbone = resnet50(weights=weights)
        in_feats = self.backbone.fc.in_features                    # 2048
        self.backbone.fc = nn.Sequential(
            nn.Linear(in_feats, 512),
            nn.ReLU(inplace=True),
            nn.Dropout(dropout),
            nn.Linear(512, n_classes),
        )

    def forward(self, x):
        return self.backbone(x)                                    # (B, n_classes)

    def param_groups(self, backbone_lr=1e-4, head_lr=1e-3, weight_decay=1e-4):
        """Return two AdamW param groups — discriminative LRs for backbone vs head."""
        head_params    = list(self.backbone.fc.parameters())
        head_param_ids = {id(p) for p in head_params}
        backbone_params = [p for p in self.backbone.parameters()
                           if id(p) not in head_param_ids]
        return [
            {"params": backbone_params, "lr": backbone_lr, "weight_decay": weight_decay},
            {"params": head_params,     "lr": head_lr,     "weight_decay": weight_decay},
        ]
```

```python
# ─────────────────────────────────────────────────────────────
# 1.3 Sanity — forward pass & parameter counts
# (pretrained=False here so the cell is fast and offline-safe;
#  the real training cells will request pretrained weights.)
# ─────────────────────────────────────────────────────────────
def _count_params(m):
    return sum(p.numel() for p in m.parameters() if p.requires_grad)

with torch.no_grad():
    dummy = torch.randn(2, 3, IMG_SIZE, IMG_SIZE)

    m1   = SmallCNN().eval()
    out1 = m1(dummy)
    print(f"SmallCNN     : output {tuple(out1.shape)}  params={_count_params(m1):,}")

    m2   = ResNet50Cls(pretrained=False).eval()
    out2 = m2(dummy)
    print(f"ResNet50Cls  : output {tuple(out2.shape)}  params={_count_params(m2):,}")

    pg = m2.param_groups()
    n_back = sum(p.numel() for p in pg[0]["params"])
    n_head = sum(p.numel() for p in pg[1]["params"])
    print(f"  backbone group : {n_back:>11,} params  (lr={pg[0]['lr']})")
    print(f"  head     group : {n_head:>11,} params  (lr={pg[1]['lr']})")

del m1, m2, dummy
```

## 1.4 Loss function — BCE-with-logits (baseline)

Multi-label classification has 20 **independent Bernoulli** outputs — an image can carry any subset of the 20 classes. The right loss is therefore the sum of 20 per-class binary cross-entropies, one per output, with a sigmoid applied to each logit:

$$L_{\text{BCE}}(y, \hat{y}) \;=\; - \sum_{c=1}^{20} \Big[\, y_c \log \sigma(\hat{y}_c) + (1 - y_c) \log\bigl(1 - \sigma(\hat{y}_c)\bigr) \,\Big]$$

We use `nn.BCEWithLogitsLoss(reduction="mean")`, which fuses the sigmoid + BCE into a single numerically stable op (uses the log-sum-exp trick — better than `torch.sigmoid` followed by `nn.BCELoss`, which can underflow for very confident predictions).

**Why not softmax + cross-entropy?** Softmax assumes mutually exclusive classes (probabilities sum to 1). For multi-label that's the wrong inductive bias — predicting "person" more strongly would mechanically force every other class score down, even when both are present in the image. Sigmoid + BCE keeps the 20 outputs independent.

This is the **baseline loss**. Section 1.X swaps it for **Asymmetric Loss (ASL)** which down-weights the dominant 92.9%-negative pairs measured in 1.1.3.

```python
# ─────────────────────────────────────────────────────────────
# 1.4 — BCE-with-logits loss for multi-label classification
# ─────────────────────────────────────────────────────────────
bce_loss_cls = nn.BCEWithLogitsLoss(reduction="mean")

# Sanity: forward + loss on one batch with a freshly-init SmallCNN.
# A random model with logits ≈ 0 gives sigmoid ≈ 0.5, so per-element
# BCE ≈ -ln(0.5) ≈ 0.693. Anything in that ballpark confirms the
# loss is wired up correctly.
with torch.no_grad():
    xb_cpu, yb_cpu = next(iter(train_loader_cls))
    m_dbg          = SmallCNN().eval()
    logits         = m_dbg(xb_cpu)
    loss_val       = bce_loss_cls(logits, yb_cpu).item()
    print(f"Logits  : {tuple(logits.shape)}  dtype={logits.dtype}")
    print(f"Targets : {tuple(yb_cpu.shape)}  pos_frac={yb_cpu.mean().item():.4f}")
    print(f"BCE loss: {loss_val:.4f}   (expected near {math.log(2):.4f} for random init)")

del m_dbg, xb_cpu, yb_cpu, logits, loss_val
```

## 1.5 Training loop & metrics

### Optimiser & schedule
- **AdamW** — Adam with decoupled weight decay (Loshchilov & Hutter 2017). Weight decay 1e-4 nudges weights toward 0; decoupling is important because plain Adam mixes weight decay into the running gradient statistics, which we don't want.
- **Cosine annealing** (`CosineAnnealingLR`, `T_max=epochs`) — starts at the initial LR and smoothly decays to 0 over the run. Empirically robust and parameter-free.
- **Discriminative LRs** for the transfer model (set in `param_groups()`): backbone 1e-4 (small — pretrained features should not move much), head 1e-3 (large — random init needs to catch up). For SmallCNN we use a single LR (1e-3) since everything is randomly initialised.
- **Mixed precision (AMP)** via `torch.cuda.amp.autocast` + `GradScaler`. ~2× speedup, ~½ memory, no measurable accuracy loss in our experience.
- **Gradient clipping** at norm 1.0 — cheap insurance against the rare early-training spike when the head is still random.

### Metrics
- **Train loss** — mean BCE per epoch.
- **Val loss** — same loss function on the held-out val (fold 0).
- **Val mAP** — `average_precision_score(..., average="macro")`. mAP integrates precision over recall and is therefore *threshold-free*; it measures the quality of the *ranking* of scores per class. We use it as the primary monitoring metric because Dice (the Kaggle metric) is threshold-dependent — comparing two epochs at threshold 0.5 would be misleading while the score distribution is still moving. Thresholds get tuned *once*, on the best-epoch's val outputs, in 1.6.

### Checkpointing
The best-mAP epoch's `state_dict` is kept in memory. After training, we restore those weights before tuning thresholds — so we evaluate the model at its peak rather than the noisier final epoch.

```python
# ─────────────────────────────────────────────────────────────
# 1.5 — train_cls + evaluate_cls
# Standard supervised loop: forward → loss → backward → step.
# Tracks train loss, val loss, and val mAP per epoch. Saves the
# state_dict from the epoch with the best val mAP so we evaluate
# at the model's peak (rather than the noisier final epoch).
# AMP is enabled by default on CUDA (~2× speedup, ~½ memory).
# ─────────────────────────────────────────────────────────────

def _amp_device(device):
    return "cuda" if torch.device(device).type == "cuda" else "cpu"


def _autocast_ctx(device, enabled):
    return autocast(_amp_device(device), enabled=enabled)


def _new_scaler(device, enabled):
    return GradScaler(_amp_device(device), enabled=enabled)


def _macro_map(targets_np, probs_np):
    aps = [
        average_precision_score(targets_np[:, c], probs_np[:, c])
        for c in range(targets_np.shape[1])
        if targets_np[:, c].sum() > 0
    ]
    return float(np.mean(aps)) if aps else float("nan")


def evaluate_cls(model, loader, loss_fn, device=DEVICE):
    """Return (mean_loss, macro_mAP, all_logits_cpu, all_targets_cpu)."""
    model.eval()
    total_loss, total_n = 0.0, 0
    all_logits, all_targets = [], []
    with torch.inference_mode():
        for xb, yb in loader:
            xb = xb.to(device, non_blocking=True)
            yb = yb.to(device, non_blocking=True)
            logits = model(xb)
            loss = loss_fn(logits, yb)
            total_loss += loss.item() * xb.size(0)
            total_n += xb.size(0)
            all_logits.append(logits.float().cpu())
            all_targets.append(yb.cpu())

    all_logits = torch.cat(all_logits)
    all_targets = torch.cat(all_targets)
    probs = torch.sigmoid(all_logits).numpy()
    targets_np = all_targets.numpy()
    return total_loss / max(total_n, 1), _macro_map(targets_np, probs), all_logits, all_targets


def train_cls(model, train_loader, val_loader, *,
              epochs, optimizer, scheduler=None, loss_fn=None,
              device=DEVICE, use_amp=True, log_every=1, name="model"):
    """Train `model`. Returns (history, best_state_dict, best_val_logits, best_val_targets)."""
    if loss_fn is None:
        loss_fn = nn.BCEWithLogitsLoss()
    model.to(device)
    use_amp = use_amp and (device.type == "cuda")
    scaler = _new_scaler(device, use_amp)

    history = {"train_loss": [], "val_loss": [], "val_mAP": []}
    best_mAP, best_state = -1.0, None

    for epoch in range(1, epochs + 1):
        model.train()
        total_loss, total_n = 0.0, 0
        for xb, yb in train_loader:
            xb = xb.to(device, non_blocking=True)
            yb = yb.to(device, non_blocking=True)
            optimizer.zero_grad(set_to_none=True)
            with _autocast_ctx(device, use_amp):
                logits = model(xb)
                loss = loss_fn(logits, yb)
            scaler.scale(loss).backward()
            scaler.unscale_(optimizer)
            torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
            scaler.step(optimizer)
            scaler.update()
            total_loss += loss.item() * xb.size(0)
            total_n += xb.size(0)

        if scheduler is not None:
            scheduler.step()

        val_loss, val_mAP, _, _ = evaluate_cls(model, val_loader, loss_fn, device=device)
        train_loss = total_loss / max(total_n, 1)
        history["train_loss"].append(train_loss)
        history["val_loss"].append(val_loss)
        history["val_mAP"].append(val_mAP)

        if val_mAP > best_mAP:
            best_mAP = val_mAP
            best_state = copy.deepcopy(model.state_dict())

        if epoch % log_every == 0:
            lr0 = optimizer.param_groups[0]["lr"]
            print(f"[{name}] ep {epoch:3d}/{epochs}  train {train_loss:.4f}  "
                  f"val {val_loss:.4f}  mAP {val_mAP:.4f}  lr {lr0:.2e}")

    model.load_state_dict(best_state)
    best_val_loss, best_val_mAP, best_logits, best_targets = evaluate_cls(
        model, val_loader, loss_fn, device=device)
    print(f"[{name}] best val mAP: {best_val_mAP:.4f}  val loss: {best_val_loss:.4f}")
    return history, best_state, best_logits, best_targets
```

## 1.6 Per-class threshold tuning + test prediction

The trained model outputs 20 raw logits per image. To produce the binary predictions Kaggle scores against, we need to:

1. Apply sigmoid → 20 probabilities ∈ [0, 1].
2. Threshold each probability to 0 / 1.

The naïve choice is a single global 0.5 threshold — but that is suboptimal for two reasons highlighted by the EDA:

- **Class-frequency skew (DA-3)** — `person` appears in 27.6% of training images while `sheep` appears in 3.6%. A model trained with BCE will systematically score rare classes lower (the calibrated *base rate* is lower), so 0.5 is too aggressive a positive-call threshold for them.
- **Severe pos/neg imbalance (DA-4)** — with 92.9% of pairs negative, even modest mis-calibration on the dominant negatives hurts global Dice.

We therefore sweep **per-class thresholds** on the val set and pick, for each class independently, the threshold that maximises that class's Dice on val. Grid: 19 values from 0.05 to 0.95. The resulting `(20,)` threshold vector is used at test time.

`predict_cls` then applies the model to the test images (using the deterministic `cls_val_tf` — same preprocessing as val), sigmoids the logits, applies the per-class thresholds, and returns an `(N_test, 20)` int array of binary predictions ready to drop into `test_df.loc[:, labels]`.


```python
# ─────────────────────────────────────────────────────────────
# 1.6 — tune_thresholds, predict_cls, run_cls_experiment
# ─────────────────────────────────────────────────────────────

def find_weights(filename, weights_dir=WEIGHTS_DIR):
    candidates = [weights_dir / filename, Path(filename)]
    return next((p for p in candidates if p.exists()), None)


def tune_thresholds(val_logits, val_targets, grid=None):
    """Vectorised per-class threshold sweep that maximises validation Dice."""
    if grid is None:
        grid = np.linspace(0.05, 0.95, 19, dtype=np.float32)
    else:
        grid = np.asarray(list(grid), dtype=np.float32)

    probs = torch.sigmoid(val_logits.float()).numpy()              # (N, C)
    targets = val_targets.numpy().astype(bool)                     # (N, C)
    preds = probs[:, :, None] >= grid[None, None, :]               # (N, C, T)

    tp = (preds & targets[:, :, None]).sum(axis=0).astype(np.float32)
    fp = (preds & ~targets[:, :, None]).sum(axis=0).astype(np.float32)
    fn = (~preds & targets[:, :, None]).sum(axis=0).astype(np.float32)
    denom = 2 * tp + fp + fn
    dice = np.divide(2 * tp, denom, out=np.zeros_like(tp), where=denom > 0)

    best_idx = dice.argmax(axis=1)
    thresholds = grid[best_idx].astype(np.float32)
    best_dices = dice[np.arange(dice.shape[0]), best_idx].astype(np.float32)
    no_positive = targets.sum(axis=0) == 0
    thresholds[no_positive] = 0.5
    best_dices[no_positive] = 0.0
    return thresholds, best_dices


def predict_cls(model, test_ds, thresholds, batch_size=64, device=DEVICE):
    """Run `model` on `test_ds`, apply per-class thresholds, return (preds, probs)."""
    loader = DataLoader(test_ds, batch_size=batch_size, shuffle=False,
                        num_workers=NUM_WORKERS, pin_memory=(device.type == "cuda"))
    model.to(device).eval()
    probs_chunks = []
    with torch.inference_mode():
        for xb in loader:
            xb = xb.to(device, non_blocking=True)
            logits = model(xb)
            probs_chunks.append(torch.sigmoid(logits).float().cpu().numpy())
    probs = np.concatenate(probs_chunks, axis=0)
    preds = (probs >= thresholds[None, :]).astype(np.int64)
    return preds, probs


def run_cls_experiment(name, model, loss_fn, epochs, weights_file, optimizer_fn,
                       scheduler_fn=None, log_every=2):
    """Shared train/load/evaluate/threshold path for all classification models."""
    model.to(DEVICE)
    if TRAIN:
        optimizer = optimizer_fn(model)
        scheduler = None if scheduler_fn is None else scheduler_fn(optimizer, epochs)
        history, best_state, val_logits, val_targets = train_cls(
            model, train_loader_cls, val_loader_cls, epochs=epochs,
            optimizer=optimizer, scheduler=scheduler, loss_fn=loss_fn,
            log_every=log_every, name=name)
        checkpoint = {"model": best_state, "history": history}
        torch.save(checkpoint, weights_file)
        print(f"Saved checkpoint → {Path(weights_file).resolve()}")
    else:
        weights_path = find_weights(weights_file)
        if weights_path is None:
            raise FileNotFoundError(
                f"{weights_file} not found in {WEIGHTS_DIR} or cwd. "
                f"Set TRAIN=True (cell 6) to train from scratch.")
        print(f"Loading weights ← {weights_path}")
        checkpoint = torch.load(weights_path, map_location=DEVICE)
        state = checkpoint.get("model", checkpoint) if isinstance(checkpoint, dict) else checkpoint
        history = checkpoint.get("history") if isinstance(checkpoint, dict) else None
        model.load_state_dict(state)
        val_loss, val_mAP, val_logits, val_targets = evaluate_cls(model, val_loader_cls, loss_fn)
        print(f"  val_loss={val_loss:.4f}  val_mAP={val_mAP:.4f}")

    thresholds, dice = tune_thresholds(val_logits, val_targets)
    print(f"{name}  mean per-class Dice (val): {dice.mean():.4f}")
    return {
        "name": name,
        "model": model,
        "history": history,
        "val_logits": val_logits,
        "val_targets": val_targets,
        "thresholds": thresholds,
        "dice": dice,
    }
```

## 1.X Improvement: Asymmetric Loss instead of plain BCE

**Motivation.** Section 1.1.3 (DA-4) measured the (image, class) pair imbalance as **7.13% positive vs 92.87% negative — a 1:13 ratio**. Plain BCE treats every positive and every negative equally, so during training the gradient is dominated by the 14× more numerous easy negatives. The model spends most of its capacity learning to confidently predict "no" rather than discriminating rare positives from confusing negatives.

**Asymmetric Loss (ASL)** [Ridnik et al., ICCV 2021 — *"Asymmetric Loss For Multi-Label Classification"*, [arXiv 2009.14119](https://arxiv.org/abs/2009.14119)] is the now-standard fix:

$$L_{\text{ASL}}(y, p) \;=\; -\sum_c \begin{cases} (1 - p_c)^{\gamma^{+}} \log(p_c) & \text{if } y_c = 1 \\ p_m^{\gamma^{-}} \log(1 - p_m) & \text{if } y_c = 0 \end{cases}$$

where $p_c = \sigma(\hat y_c)$ is the sigmoid probability and $p_m = \max(p_c - m, 0)$ is the probability shifted down by margin $m$ (a "probability clip" that lets the model treat confidently-negative samples as fully negative).

**Two key knobs**, both tuned to the negative class:
- **$\gamma^- = 4 \gg \gamma^+ = 1$** — the focal exponent on negatives is 4× the one on positives, so easy negatives are down-weighted *much* more aggressively than easy positives. This rebalances the gradient so the model isn't drowning in easy "no"s.
- **$m = 0.05$** — anything the model already thinks has probability ≤ 5% of being positive contributes *zero* to the loss. Stops the model from squeezing additional negatives below an already-low probability (a wasted gradient).

ASL is a drop-in replacement for `BCEWithLogitsLoss` (same logits-in, scalar-out interface) and inherits its numerical-stability tricks. We keep everything else identical to 1.8 — same architecture, optimiser, schedule, augmentations — so the *only* difference between 1.8 and 1.9 is the loss. That's a clean ablation: any difference in val Dice is attributable to ASL.

**Where we expect to see the gain.** Rare classes (`sheep`, `cow` at 3-6%, see 1.1.2) — those are the ones where the negative-dominance was most punishing for plain BCE. Frequent classes (`person` at 27.6%) are already well-served by BCE.


```python
# ─────────────────────────────────────────────────────────────
# 1.X — AsymmetricLossOptimized (Ridnik et al., ICCV 2021)
# Drop-in replacement for BCEWithLogitsLoss with two extras:
#   • Asymmetric focal: γ⁻=4 (negatives) ≫ γ⁺=1 (positives)
#   • Probability margin m=0.05: shifts negative probabilities by m and
#     clips, so an "already confidently negative" sample (p ≤ m)
#     contributes zero to the loss.
# ─────────────────────────────────────────────────────────────

class AsymmetricLossOptimized(nn.Module):
    """ASL with reduction='mean' (matches BCEWithLogitsLoss's interface)."""
    def __init__(self, gamma_neg=4, gamma_pos=1, clip=0.05, eps=1e-8):
        super().__init__()
        self.gamma_neg = gamma_neg
        self.gamma_pos = gamma_pos
        self.clip      = clip
        self.eps       = eps

    def forward(self, logits, targets):
        # Probabilities
        xs_pos = torch.sigmoid(logits)
        xs_neg = 1.0 - xs_pos

        # Asymmetric clipping (probability margin for negatives)
        if self.clip > 0:
            xs_neg = (xs_neg + self.clip).clamp(max=1.0)

        # Basic BCE
        loss = (targets       * torch.log(xs_pos.clamp(min=self.eps)) +
                (1 - targets) * torch.log(xs_neg.clamp(min=self.eps)))

        # Asymmetric focal weighting (γ⁻ on negatives, γ⁺ on positives)
        if self.gamma_neg > 0 or self.gamma_pos > 0:
            xs_pos_w = xs_pos * targets
            xs_neg_w = xs_neg * (1 - targets)
            asymmetric_w = torch.pow(
                1 - xs_pos_w - xs_neg_w,
                self.gamma_pos * targets + self.gamma_neg * (1 - targets))
            loss = loss * asymmetric_w

        return -loss.mean()


# Default ASL with Ridnik's recommended hyperparameters
asl_loss_cls = AsymmetricLossOptimized(gamma_neg=4, gamma_pos=1, clip=0.05)

# Sanity: BCE vs ASL on a freshly-init random batch.
# At init (logits ≈ 0), BCE ≈ 0.69. ASL is much smaller because the focal
# weight (1-p)^γ is < 1 and the negative term is further suppressed by γ⁻=4.
with torch.no_grad():
    set_seed()
    xb_cpu, yb_cpu = next(iter(train_loader_cls))
    m_dbg          = SmallCNN().eval()
    logits         = m_dbg(xb_cpu)
    bce_v          = bce_loss_cls(logits, yb_cpu).item()
    asl_v          = asl_loss_cls(logits, yb_cpu).item()
    print(f"BCE loss (random init): {bce_v:.4f}")
    print(f"ASL loss (random init): {asl_v:.4f}   (smaller — easy negatives suppressed)")

del m_dbg, xb_cpu, yb_cpu, logits
```

## 1.11 Performance & error analysis

The aggregate numbers (mAP, mean Dice) hide the most interesting question: **which classes does each model get right or wrong, and why?** This subsection breaks the val performance down by class for all three models, focuses on where ASL helps vs BCE, then inspects the hardest failure cases.

We compute per-class:
- **Precision** $= TP / (TP + FP)$ — of images we *say* contain class $c$, how many do.
- **Recall** $= TP / (TP + FN)$ — of images that *do* contain class $c$, how many we caught.
- **F1 / Dice** $= 2\,TP / (2\,TP + FP + FN)$ — harmonic mean of precision and recall (this is the same per-class Dice we tuned thresholds against).

All metrics are at the **per-class tuned threshold from 1.6**, so each class is at its own optimal operating point on the precision-recall trade-off.

**The expectation from the EDA**: ASL's gains should land on the **rare classes** (`sheep`, `cow`, `bus` at ~3-5% of the training set), where BCE's negative-gradient dominance was most punishing. Frequent classes (`person` at 27.6%) should be roughly tied between BCE and ASL since BCE was already well-served there.


```python
# ─────────────────────────────────────────────────────────────
# 1.11.1 — Per-class precision / recall / F1 for all three models
# ─────────────────────────────────────────────────────────────

def per_class_metrics(logits, targets, thresholds):
    """Per-class P/R/F1 at the supplied thresholds. Returns list of dicts."""
    probs = torch.sigmoid(logits.float()).numpy()
    targets_np = targets.numpy().astype(bool)
    preds = probs >= thresholds[None, :]

    tp = (preds & targets_np).sum(axis=0).astype(np.float32)
    fp = (preds & ~targets_np).sum(axis=0).astype(np.float32)
    fn = (~preds & targets_np).sum(axis=0).astype(np.float32)

    prec = np.divide(tp, tp + fp, out=np.zeros_like(tp), where=(tp + fp) > 0)
    rec = np.divide(tp, tp + fn, out=np.zeros_like(tp), where=(tp + fn) > 0)
    f1 = np.divide(2 * tp, 2 * tp + fp + fn, out=np.zeros_like(tp), where=(2 * tp + fp + fn) > 0)

    return [
        {"precision": float(prec[c]), "recall": float(rec[c]), "f1": float(f1[c]),
         "tp": int(tp[c]), "fp": int(fp[c]), "fn": int(fn[c])}
        for c in range(len(thresholds))
    ]

m_small = per_class_metrics(val_logits_smallcnn,   val_targets_smallcnn,   thresholds_smallcnn)
m_bce   = per_class_metrics(val_logits_resnet_bce, val_targets_resnet_bce, thresholds_resnet_bce)
m_asl   = per_class_metrics(val_logits_resnet_asl, val_targets_resnet_asl, thresholds_resnet_asl)

# ── Bar chart: per-class F1 for all three models ──
fig, ax = plt.subplots(figsize=(16, 5))
x = np.arange(len(labels))
w = 0.27
ax.bar(x - w, [m["f1"] for m in m_small], width=w, label="SmallCNN",     color="lightcoral", alpha=0.85)
ax.bar(x    , [m["f1"] for m in m_bce  ], width=w, label="ResNet50+BCE", color="steelblue",  alpha=0.85)
ax.bar(x + w, [m["f1"] for m in m_asl  ], width=w, label="ResNet50+ASL", color="seagreen",   alpha=0.85)
ax.set_xticks(x); ax.set_xticklabels(labels, rotation=55)
ax.set_ylabel("Per-class F1 (val, tuned threshold)")
ax.set_title("Per-class F1 — three models compared")
ax.set_ylim(0, 1.05)
ax.grid(axis="y", alpha=0.3)
ax.legend(loc="lower right")
plt.tight_layout(); plt.show()
```

# 2. Semantic segmentation
The goal here is to implement a segmentation model that labels every pixel in the image as belonging to one of the 20 classes (and/or background). Use the training set to train your model and compete on the test set (by filling in the segmentation column in the test dataframe).

```python
class RandomSegmentationModel:
    """
    Random segmentation model: 
        - generates random label maps for the inputs based on the class distributions observed during training
        - every pixel in an input can only have one label
    """
    def fit(self, X, Y):
        """
        Adjusts the class ratio variable to the one observed in Y. 

        Parameters
        ----------
        X: list of arrays - n x (height x width x 3)
        Y: list of arrays - n x (height x width)

        Returns
        -------
        self
        """
        self.distribution = np.mean([[np.sum(Y_ == i) / Y_.size for i in range(len(labels) + 1)] for Y_ in Y], axis=0)
        print("Setting class distribution to:\nbackground: {}\n{}".format(self.distribution[0], "\n".join(f"{label}: {p}" for label, p in zip(labels, self.distribution[1:]))))
        return self
        
    def predict(self, X):
        """
        Predicts for each input a label map.
        
        Parameters
        ----------
        X: list of arrays - n x (height x width x 3)
            
        Returns
        -------
        Y_pred: list of arrays - n x (height x width)
        """
        np.random.seed(0)
        return [np.random.choice(np.arange(len(labels) + 1), size=X_.shape[:2], p=self.distribution) for X_ in X]
    
    def __call__(self, X):
        return self.predict(X)
    
model = RandomSegmentationModel()
model.fit(train_df["img"], train_df["seg"])
test_df.loc[:, "seg"] = model.predict(test_df["img"])
test_df.head(1)
```

# 3. Adversarial attack

We focus on a **white-box targeted classification attack** against our trained CLIP multi-label classifier. The segmentation models above are useful for the assignment, but their validation Dice is still low in this run, so attacking the classifier gives a cleaner experiment: we can tell whether the adversary fooled a model that has learned a meaningful signal.

**Task.** We train the perturbation network on classification-training images where the ground-truth label does **not** contain `aeroplane`, then evaluate and visualise it on validation images where the target class is absent. The validation images act as test-style examples because we know `aeroplane` is not present; the same perturbation generator can be applied to the unlabeled Kaggle test images.

**Victim and adversary.** The victim is the already-trained `model_clip`. Its parameters are frozen. The adversary is a small encoder-decoder `f_adv(x; theta_adv)` that outputs an RGB perturbation `delta`. We add it to the clean image in pixel space and clip the result back to the valid image range:

`x_adv = clip(x + delta, 0, 1)`, with `||delta||_inf <= epsilon` and `epsilon = 8/255`.

**Losses.** The task loss is a targeted binary cross-entropy on the `aeroplane` logit:

`L_task = BCEWithLogits(h_cls(x_adv)[aeroplane], 1)`.

The regularisation keeps the perturbation small and visually unobtrusive:

`R = lambda_l2 * mean(delta^2) + lambda_tv * TV(delta)`.

The full objective is `L_task + R`. Only `theta_adv` is optimized; `theta_cls` stays unchanged.


```python
# White-box targeted attack on the CLIP multi-label classifier.
# Run the CLIP section first so that model_clip, thresholds_clip, and val_df_cls exist.
TARGET_CLASS = "aeroplane"
TARGET_IDX = list(labels).index(TARGET_CLASS)

ATTACK_EPS = 8 / 255          # maximum per-channel pixel perturbation
ATTACK_BATCH_SIZE = min(16, CLIP_BATCH_SIZE)
ATTACK_EPOCHS = 8 if not SMOKE else 1
ATTACK_LR = 2e-3
ATTACK_L2_WEIGHT = 10.0
ATTACK_TV_WEIGHT = 0.25
ATTACK_NUM_WORKERS = 0 if os.name == "nt" else NUM_WORKERS

clip_mean = torch.tensor([0.485, 0.456, 0.406], device=device).view(1, 3, 1, 1)
clip_std = torch.tensor([0.229, 0.224, 0.225], device=device).view(1, 3, 1, 1)


def denormalize_clip_batch(x_norm):
    return x_norm * clip_std + clip_mean


def normalize_clip_batch(x_pixel):
    return (x_pixel - clip_mean) / clip_std


def total_variation(delta):
    tv_h = (delta[:, :, 1:, :] - delta[:, :, :-1, :]).abs().mean()
    tv_w = (delta[:, :, :, 1:] - delta[:, :, :, :-1]).abs().mean()
    return tv_h + tv_w


class PerturbationGenerator(nn.Module):
    """Small encoder-decoder that predicts a bounded RGB perturbation."""
    def __init__(self, hidden=32):
        super().__init__()
        self.net = nn.Sequential(
            nn.Conv2d(3, hidden, kernel_size=3, padding=1, bias=False),
            nn.BatchNorm2d(hidden),
            nn.ReLU(inplace=True),
            nn.Conv2d(hidden, hidden * 2, kernel_size=3, stride=2, padding=1, bias=False),
            nn.BatchNorm2d(hidden * 2),
            nn.ReLU(inplace=True),
            nn.Conv2d(hidden * 2, hidden * 2, kernel_size=3, padding=1, bias=False),
            nn.BatchNorm2d(hidden * 2),
            nn.ReLU(inplace=True),
            nn.ConvTranspose2d(hidden * 2, hidden, kernel_size=4, stride=2, padding=1, bias=False),
            nn.BatchNorm2d(hidden),
            nn.ReLU(inplace=True),
            nn.Conv2d(hidden, 3, kernel_size=3, padding=1),
            nn.Tanh(),
        )

    def forward(self, x_norm):
        return self.net(x_norm)


def make_adversarial_batch(x_norm, adversary, eps=ATTACK_EPS):
    x_pixel = denormalize_clip_batch(x_norm).clamp(0, 1)
    raw_delta = eps * adversary(x_norm)
    x_adv_pixel = (x_pixel + raw_delta).clamp(0, 1)
    delta = x_adv_pixel - x_pixel
    x_adv_norm = normalize_clip_batch(x_adv_pixel)
    return x_adv_norm, x_adv_pixel, delta, x_pixel


def train_targeted_adversary(victim, adversary, loader, optimizer, target_idx, threshold):
    history = []
    victim.eval()

    for epoch in range(1, ATTACK_EPOCHS + 1):
        adversary.train()
        totals = {"n": 0, "loss": 0.0, "task": 0.0, "l2": 0.0, "tv": 0.0, "success": 0.0}

        for xb, yb in loader:
            negative_mask = yb[:, target_idx] < 0.5
            if not negative_mask.any():
                continue

            xb = xb[negative_mask].to(device, non_blocking=True)
            batch_n = xb.size(0)

            optimizer.zero_grad(set_to_none=True)
            x_adv_norm, _, delta, _ = make_adversarial_batch(xb, adversary)
            logits_adv = victim(x_adv_norm)

            target = torch.ones_like(logits_adv[:, target_idx])
            task_loss = F.binary_cross_entropy_with_logits(logits_adv[:, target_idx], target)
            l2_loss = delta.pow(2).mean()
            tv_loss = total_variation(delta)
            loss = task_loss + ATTACK_L2_WEIGHT * l2_loss + ATTACK_TV_WEIGHT * tv_loss

            loss.backward()
            torch.nn.utils.clip_grad_norm_(adversary.parameters(), max_norm=1.0)
            optimizer.step()

            with torch.no_grad():
                probs_adv = torch.sigmoid(logits_adv[:, target_idx])
                totals["n"] += batch_n
                totals["loss"] += loss.item() * batch_n
                totals["task"] += task_loss.item() * batch_n
                totals["l2"] += l2_loss.item() * batch_n
                totals["tv"] += tv_loss.item() * batch_n
                totals["success"] += (probs_adv > threshold).float().sum().item()

        if totals["n"] == 0:
            raise ValueError(f"No validation images without class {TARGET_CLASS!r} were found.")

        summary = {
            "epoch": epoch,
            "loss": totals["loss"] / totals["n"],
            "task_loss": totals["task"] / totals["n"],
            "l2": totals["l2"] / totals["n"],
            "tv": totals["tv"] / totals["n"],
            "target_rate": totals["success"] / totals["n"],
        }
        history.append(summary)
        print(
            f"Attack epoch {epoch:02d}/{ATTACK_EPOCHS} | "
            f"loss {summary['loss']:.4f} | task {summary['task_loss']:.4f} | "
            f"target-rate {summary['target_rate']:.3f}"
        )

    return pd.DataFrame(history)


@torch.no_grad()
def evaluate_targeted_adversary(victim, adversary, loader, target_idx, threshold):
    victim.eval()
    adversary.eval()
    totals = {
        "n": 0,
        "clean_target": 0.0,
        "adv_target": 0.0,
        "clean_prob": 0.0,
        "adv_prob": 0.0,
        "linf": 0.0,
        "l2": 0.0,
    }

    for xb, yb in loader:
        negative_mask = yb[:, target_idx] < 0.5
        if not negative_mask.any():
            continue

        xb = xb[negative_mask].to(device, non_blocking=True)
        batch_n = xb.size(0)

        logits_clean = victim(xb)
        x_adv_norm, _, delta, _ = make_adversarial_batch(xb, adversary)
        logits_adv = victim(x_adv_norm)

        clean_probs = torch.sigmoid(logits_clean[:, target_idx])
        adv_probs = torch.sigmoid(logits_adv[:, target_idx])

        totals["n"] += batch_n
        totals["clean_target"] += (clean_probs > threshold).float().sum().item()
        totals["adv_target"] += (adv_probs > threshold).float().sum().item()
        totals["clean_prob"] += clean_probs.sum().item()
        totals["adv_prob"] += adv_probs.sum().item()
        totals["linf"] += delta.abs().flatten(1).max(dim=1).values.sum().item()
        totals["l2"] += delta.flatten(1).norm(p=2, dim=1).sum().item()

    if totals["n"] == 0:
        raise ValueError(f"No validation images without class {TARGET_CLASS!r} were found.")

    n = totals["n"]
    return {
        "target_class": TARGET_CLASS,
        "n_clean_without_target": int(n),
        "threshold": threshold,
        "clean_target_rate": totals["clean_target"] / n,
        "adv_target_rate": totals["adv_target"] / n,
        "mean_clean_prob": totals["clean_prob"] / n,
        "mean_adv_prob": totals["adv_prob"] / n,
        "mean_linf": totals["linf"] / n,
        "mean_l2": totals["l2"] / n,
    }


if Path("clip_vitb16_multilabel.pt").exists():
    model_clip.load_state_dict(torch.load("clip_vitb16_multilabel.pt", weights_only=True, map_location=device))
model_clip = model_clip.to(device).eval()
for param in model_clip.parameters():
    param.requires_grad_(False)

attack_threshold = float(thresholds_clip[TARGET_IDX]) if "thresholds_clip" in globals() else 0.5
_, attack_val_tf = get_transforms("clip")
attack_train_dataset = VOCClassificationDataset(train_df_cls, labels, attack_val_tf)
attack_eval_dataset = VOCClassificationDataset(val_df_cls, labels, attack_val_tf)
attack_loader = DataLoader(
    attack_train_dataset,
    batch_size=ATTACK_BATCH_SIZE,
    shuffle=True,
    num_workers=ATTACK_NUM_WORKERS,
    pin_memory=(device.type == "cuda"),
)
attack_eval_loader = DataLoader(
    attack_eval_dataset,
    batch_size=ATTACK_BATCH_SIZE,
    shuffle=False,
    num_workers=ATTACK_NUM_WORKERS,
    pin_memory=(device.type == "cuda"),
)

adversary = PerturbationGenerator(hidden=32).to(device)
attack_optimizer = torch.optim.AdamW(adversary.parameters(), lr=ATTACK_LR, weight_decay=1e-5)

print(f"Targeted white-box attack: force '{TARGET_CLASS}' on images where it is absent.")
print(f"Decision threshold for {TARGET_CLASS!r}: {attack_threshold:.3f}")
attack_history = train_targeted_adversary(model_clip, adversary, attack_loader, attack_optimizer, TARGET_IDX, attack_threshold)
attack_report = evaluate_targeted_adversary(model_clip, adversary, attack_eval_loader, TARGET_IDX, attack_threshold)

print("\nAttack report")
print(pd.Series(attack_report))
```

```python
@torch.no_grad()
def visualize_targeted_attack_examples(victim, adversary, loader, target_idx, target_class, threshold, num_examples=4):
    victim.eval()
    adversary.eval()
    examples = []

    for xb, yb in loader:
        negative_mask = yb[:, target_idx] < 0.5
        if not negative_mask.any():
            continue

        yb_clean = yb[negative_mask].cpu()
        xb = xb[negative_mask].to(device, non_blocking=True)

        logits_clean = victim(xb)
        x_adv_norm, x_adv_pixel, delta, x_pixel = make_adversarial_batch(xb, adversary)
        logits_adv = victim(x_adv_norm)

        clean_probs = torch.sigmoid(logits_clean[:, target_idx]).cpu().numpy()
        adv_probs = torch.sigmoid(logits_adv[:, target_idx]).cpu().numpy()

        for i in range(xb.size(0)):
            true_labels = [labels[j] for j in np.where(yb_clean[i].numpy() == 1)[0]]
            examples.append({
                "clean": x_pixel[i].cpu().permute(1, 2, 0).numpy(),
                "delta": delta[i].cpu().permute(1, 2, 0).numpy(),
                "adv": x_adv_pixel[i].cpu().permute(1, 2, 0).numpy(),
                "clean_prob": clean_probs[i],
                "adv_prob": adv_probs[i],
                "true_labels": true_labels,
            })
            if len(examples) >= num_examples:
                break

        if len(examples) >= num_examples:
            break

    if not examples:
        raise ValueError(f"No examples without class {target_class!r} were available for visualization.")

    rows = len(examples)
    fig, axs = plt.subplots(rows, 3, figsize=(12, 3.6 * rows))
    if rows == 1:
        axs = np.expand_dims(axs, axis=0)

    for row, ex in enumerate(examples):
        delta_vis = np.clip(ex["delta"] / (2 * ATTACK_EPS) + 0.5, 0, 1)
        label_text = ", ".join(ex["true_labels"][:4]) if ex["true_labels"] else "none"

        axs[row, 0].imshow(np.clip(ex["clean"], 0, 1))
        axs[row, 0].set_title(f"Clean\nP({target_class})={ex['clean_prob']:.3f}\nTrue: {label_text}", fontsize=9)
        axs[row, 0].axis("off")

        axs[row, 1].imshow(delta_vis)
        axs[row, 1].set_title(f"Perturbation\nscaled around 0.5", fontsize=9)
        axs[row, 1].axis("off")

        axs[row, 2].imshow(np.clip(ex["adv"], 0, 1))
        axs[row, 2].set_title(f"Perturbed\nP({target_class})={ex['adv_prob']:.3f}\nthreshold={threshold:.3f}", fontsize=9)
        axs[row, 2].axis("off")

    plt.tight_layout()
    plt.show()


visualize_targeted_attack_examples(
    model_clip,
    adversary,
    attack_eval_loader,
    TARGET_IDX,
    TARGET_CLASS,
    attack_threshold,
    num_examples=4,
)
```

```python
## 3.1 Reflection on the attack

**Was the adversary successful?** The printed report compares `clean_target_rate` with `adv_target_rate` on images that do not contain the target class. The attack is successful when `adv_target_rate` is clearly higher while `mean_linf` remains close to the enforced `8/255` bound. This is the right comparison because the clean rate is just the classifier's normal false-positive rate for `aeroplane`, while the adversarial rate measures the new false positives caused by the learned perturbation.

**Are the images still recognisable?** The visual grid shows the original image, the perturbation, and the perturbed image. Because the perturbation is clipped to `epsilon = 8/255` per RGB channel and additionally regularised by L2 and total variation terms, the perturbed images should remain recognisable to a human observer. The perturbation panel is deliberately rescaled so that we can see it; it is not displayed at its true visual strength.

**Is this realistic?** This is realistic for a digital white-box setting where an attacker knows the exact model, preprocessing, and weights, for example when a model is shipped inside an app or exposed in a reproducible open-source pipeline. It is less realistic as a physical-world attack: printing, camera noise, lighting, resizing, compression, and viewpoint changes can destroy a perturbation trained only for this exact input pipeline. A stronger real-world version would train with expectation over transformations and test under camera-like corruptions.

**Does this make the initial model unreliable?** It does not mean every prediction from the classifier is useless, but it does show that the model has a local vulnerability: small, structured changes can move the decision boundary in a way humans do not expect. Therefore, high validation performance alone is not enough evidence for safety under adversarial conditions.

**How could robustness be improved?** General options include adversarial training with FGSM/PGD-style examples, augmentations that simulate real acquisition changes, expectation-over-transformations training for physical robustness, input denoising or compression checks, monitoring confidence/OOD behaviour, using ensembles, and evaluating the model under attacks as part of validation rather than only on clean data.
```
